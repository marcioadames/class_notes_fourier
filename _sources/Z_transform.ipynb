# Códigos para calcular a transformada Z

```python
# method using lcapy

import lcapy as lc
from lcapy.discretetime import n

f_n=n*2**n*lc.exp(-3*n)
Z_f=f_n.ZT()
print(Z_f)
```

```python
#method using sympy and tbcontrol

import sympy as sym
import tbcontrol

z = sym.symbols('z')
n = sym.symbols('n', integer=True)

f_n=n*2**n*sym.exp(-3*n)

Z_f = sym.Sum(f_n, (n, 0, sym.oo))
Z_f.doit()
```
