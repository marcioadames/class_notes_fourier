# Transformada de Discreta Fourier (DFT)

```python
from numpy import *
import matplotlib.pyplot as plt

# intervalo de interesse (-a,a)
a = 10 
numpoints = 101
x = linspace(0,a, numpoints, endpoint = True, dtype = complex)

# Definindo um sinal randômico
noise_freq = 3
coef = 10*random.rand(3,1) 
coeff =  random.rand(noise_freq,1)
r = 100*random.rand(3,1)
rr = 100*random.rand(noise_freq,1)

f = coef[0]*sin(r[0]*x*a) + coef[1]*sin(r[1]*x*a) + coef[2]*sin(r[2]*x*a)

for i in range(noise_freq):
    f = f+coeff[i]*sin(rr[i]*x*a)


# Calculando a DFT
w = linspace(-a,a,numpoints,endpoint = True)
F_f = zeros(len(w), dtype = complex)
for i in range(len(F_f)):
    for j in range(len(F_f)):
        F_f[i] = F_f[i]+f[j]*exp(-1j*2*j*pi*i/numpoints)
F_f = F_f/len(x)


# removendo frequencias pequenas
"""
cut = 1
for i in range(len(F_f)):
    if abs(F_f[i])< cut:
        F_f[i] = 0 

#"""

# Calculando a transformada inversa

ff = zeros(len(w), dtype = complex)
for i in range(len(ff)):
    for j in range(len(F_f)):
        ff[i] = ff[i]+F_f[j]*exp(1j*2*j*pi*i/numpoints)
ff = ff


# Plotando a função
plt.plot(x, real(f))

# Plotando a transformada
plt.plot(x, real(F_f))

# Tentando plotar a transformada INVERSA
plt.plot(x, real(ff))

plt.show()
```
