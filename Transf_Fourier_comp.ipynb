# Transformada de Fourier complexa simbólica

```python
# import fourier_transform
from sympy import *
from sympy.abc import x, w

# Definindo uma função simbólica
f = exp(-x**2)

print(f)

#f = Piecewise( ( 1, (x > -1) & (x < 1) ), (0, True))

#f = Piecewise( ( 1, (x > 0)), (0, True))

# Calculando a Transformada de Fourier
F_f = fourier_transform(f, x, w)

print(F_f)

# Calculando a transformada inversa

ff = inverse_fourier_transform(F_f, w,x)

print(ff)

# o símbolo "|" significa "ou"
```

```python
# import fourier_transform
from sympy import *
from sympy.abc import x, w

# Definindo uma função simbólica
f = exp(-x**2)

print(f)

#f = Piecewise( ( 1, (x > -1) & (x < 1) ), (0, True))

#f = Piecewise( ( 1, (x > 0)), (0, True))

# Calculando a Transformada de Fourier
F_f = fourier_transform(f, x, w)

# Calculando a transformada inversa

ff = inverse_fourier_transform(F_f, w,x)


# Plotando a função
plot(f)

# Plotando a transformada
plot(F_f)

# Tentando plotar a transformada INVERSA
plot(ff)
```
